```python
import cv2 
import matplotlib.pyplot as plt
import numpy as np
```

```python
img = cv2.imread('sudoku_2.png')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
gray=cv2.copyMakeBorder(gray, 15, 15, 15, 15, cv2.BORDER_CONSTANT, 0)
plt.imshow(gray, cmap='gray')
```

```python
edges_Canny = cv2.Canny(gray, 30, 150)
plt.imshow(edges_Canny, cmap='gray')
plt.title('Edges Detected')
plt.show()
```

```python
indices = np.argwhere(edges_Canny != 0)

first_pos = indices[0]
last_pos = indices[-1]
h = last_pos[0] - first_pos[0]
w = last_pos[1] - first_pos[1]
one_cell_h = h // 9 + 2
one_cell_w = w // 9 + 2
print(first_pos, last_pos)
```

```python
for i in range(9):
    for j in range(9):
        cell = gray[first_pos[0]+i*one_cell_h:first_pos[0]+(i+1)*one_cell_h, 
                           first_pos[1]+j*one_cell_w:first_pos[1]+(j+1)*one_cell_w]
        cv2.imwrite(f"sudoku_3_{i}_{j}.jpg", cell)
```
